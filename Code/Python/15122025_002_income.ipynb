```python
# df11 = pd.read_csv("madeData/japan_regions.csv")
df11 = pd.read_csv('/Users/sus/Library/CloudStorage/OneDrive-UniversityofGeorgia/1 UGA/1 PhD/1 Spring24/Tobacco/madeData/pop_income_japan.csv')

japan_pop_income_df_shape = df11.shape
# df11.info()

## Income is in million yen

```

```python
## firstly I need to separate the spaces and the string

df11['inc_2017'] = df11['inc_2017'].str.replace(',', '').astype(int)
df11['inc_2018'] = df11['inc_2018'].str.replace(',', '').astype(int)
df11['inc_2019'] = df11['inc_2019'].str.replace(',', '').astype(int)
df11['inc_2020'] = df11['inc_2020'].str.replace(',', '').astype(int)
df11['pop_2017'] = df11['pop_2017'].str.replace(',', '').astype(int)
df11['pop_2018'] = df11['pop_2018'].str.replace(',', '').astype(int)
df11['pop_2019'] = df11['pop_2019'].str.replace(',', '').astype(int)
df11['pop_2020'] = df11['pop_2020'].str.replace(',', '').astype(int)

df11.info()
```

```python
df11['2017_pop'] = df11.groupby(['region_id'])['pop_2017'].transform('sum')
df11['2018_pop'] = df11.groupby(['region_id'])['pop_2018'].transform('sum')
df11['2019_pop'] = df11.groupby(['region_id'])['pop_2019'].transform('sum')
df11['2020_pop'] = df11.groupby(['region_id'])['pop_2020'].transform('sum')
df11['2017_inc'] = df11.groupby(['region_id'])['inc_2017'].transform('sum')
df11['2018_inc'] = df11.groupby(['region_id'])['inc_2018'].transform('sum')
df11['2019_inc'] = df11.groupby(['region_id'])['inc_2019'].transform('sum')
df11['2020_inc'] = df11.groupby(['region_id'])['inc_2020'].transform('sum')


df11 = df11[['region_id', '2017_pop', '2018_pop', '2019_pop', '2020_pop', '2017_inc',	'2018_inc',	'2019_inc',	'2020_inc']].copy()
df11
```

```python
df11 = df11.drop_duplicates()
japan_pop_income_df_dedup_shape = df11.shape
df11
```

```python
# lets melt the dataset to make it the long form
# df11 = df11['region_id', 'region_name', '2017_pop', '2018_pop', '2019_pop',	'2020_pop', '2017_inc',	'2018_inc',	'2019_inc',	'2020_inc'].drop_duplicates().copy()
df_long = pd.melt(df11, id_vars=['region_id'], value_vars=['2017_pop',	'2018_pop',	'2019_pop',	'2020_pop'], var_name='PopYear', value_name='PopValue')
df_long2 = pd.melt(df11, id_vars=['region_id'], value_vars=['2017_inc',	'2018_inc',	'2019_inc',	'2020_inc'], var_name='IncYear', value_name='IncValue')
# df_long['PopYear'] = 
df_concat = pd.concat([df_long, df_long2['IncValue']], axis="columns")
df_concat 

# df_long = df_long.rename(columns = {'PopYear: Year'})
# merged_df = pd.merge(df, df1, on=['region', 'year'], how='inner')

df_concat['year'] = df_concat['PopYear'].str.split('[ _]').str[0]
```

```python
df1 = df_concat[['region_id', 'year', 'PopValue', 'IncValue']]
df1 = df1.rename(columns = {'region_id': 'region'})
```

```python
def recode_total_as13(region):
    if region == 0:
        return 13
    return region

df1['region'] = df1.region.apply(recode_total_as13)
df1
```

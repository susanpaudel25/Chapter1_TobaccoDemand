```python
# Loading the dataset
df = pd.read_excel('monthly_CPI.xlsx')

df # region week level

# Loading the dataset
df2 = pd.read_csv("FINAL_Chapter1.csv")

df

df2
```

```python
# merging these two datasets for getting the other variables like price indices and monthly budget share and quantities

merged_df1 = pd.merge(df2, df, on=[ 'region', 'W'], how='inner')

merged_df1
```

```python
merged_df22 = merged_df1[['month', 'region', 'CPI_x', 'ln_CPI_x', 'totrev_Smoke', 'totrev_Heat', 'qt_Smoke', 'qt_Heat', 'Tornqvist_Smoke', 'Tornqvist_Heat', 'PopValue' ]]
merged_df22
```

```python
# 2. Aggregate to Monthly Level (Collapsing the data)
# We group by Region, Year, and Month to preserve the panel structure
monthly_df = merged_df22.groupby(['region', 'month']).agg({
    'totrev_Smoke': 'sum',         # Sum expenditures
    'totrev_Heat': 'sum',
    'qt_Smoke': 'sum',             # Sum quantities (packs)
    'qt_Heat': 'sum',   # Average income (Income flow is smooth)
    'Tornqvist_Smoke': 'mean',
    'Tornqvist_Heat': 'mean',
    'CPI_x': 'mean',
    'ln_CPI_x': 'mean',
    'PopValue': 'mean'                  # Average CPI (Price index is smooth)
}).reset_index()
```

```python

monthly_df
```

```python

# Loading the dataset
df22 = pd.read_excel("final_inc_exp.xlsx")


# We group by Region, Year, and Month to preserve the panel structure
monthly_df2 = df22.groupby(['region', 'month']).agg({
    'monthly_percap_exp': 'mean',         # Sum expenditures
    'monthly_percap_income': 'mean'                 # Average CPI (Price index is smooth)
}).reset_index()


monthly_df2
```

```python
# merge this with the monthly income or expenditure variables
merged_df = pd.merge(monthly_df, monthly_df2, on=[ 'region', 'month'], how='inner')

merged_df
```
